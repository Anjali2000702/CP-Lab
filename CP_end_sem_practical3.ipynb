```python
import cupy as cp

print("CuPy Version:", cp.__version__)
```

```python
import numpy as np

N = 1_000_000  # 1 Million elements

data = np.random.randint(
    0,
    1000000,
    size=N,
    dtype=np.int32
)

print("Dataset Size:", len(data))
print("First 10 values:")
print(data[:10])
```

```python
import cupy as cp

gpu_data = cp.asarray(data)

print("Data transferred to GPU")
print("GPU Array Shape:", gpu_data.shape)
```

```python
import time

start_gpu = time.time()

gpu_sorted = cp.sort(gpu_data)

cp.cuda.Stream.null.synchronize()

end_gpu = time.time()

gpu_time = end_gpu - start_gpu

print("GPU Sorting Time:", gpu_time, "seconds")
```

```python
sorted_cpu = cp.asnumpy(gpu_sorted)

print("First 10 Sorted Values:")
print(sorted_cpu[:10])

print("\nLast 10 Sorted Values:")
print(sorted_cpu[-10:])

print("\nArray Sorted:",
      np.all(sorted_cpu[:-1] <= sorted_cpu[1:]))
```

```python
import time

start_cpu = time.time()

cpu_sorted = np.sort(data)

end_cpu = time.time()

cpu_time = end_cpu - start_cpu

print("CPU Sorting Time:", cpu_time, "seconds")
```
